```python
#Q7
class Rectangle:
#define the rectangle with its length and width properties
  def __init__(self, length, width):
    self.length = length
    self.width = width
#Calculate the area of the rectangle
  def calculate_area(self):
    return self.length * self.width
#Example
rect = Rectangle(51, 3)
area = rect.calculate_area()
print("The area of rectangle is:",area) #display the area of the rectangle
```

```python
#Q8
#Write a program that uses a while loop to calculate the sum of all numbers from 1 to 10
total = 0
number = 1
#adds up each number one by one in while loop till 10 to the total value
while number <= 10:
  total += number
  number += 1
print("The sum of numbers from 1 to 10 is:", total) #display the output value
```

```python
#Q9
product = 1
print("Enter numbers one by one and type 'done' when finished:")
#Loop starts by asking the user input in numbers till 'done' is typed
while True:
  user_input = input("Enter a number (or 'done' to finish) : ")
  if user_input == 'done':
    break #breaks the loop if user input is 'done'
  try:
    number = float(user_input)
    product *= number #multiplies each input numbers
  except ValueError: #if input value is invalid it raise the value error
    print("Invalid input. Please enter a number or 'done'.")
print("The product of the entered numbers is:", product)
```

```python
#importing seaborn and pandas packages
import seaborn as sns
import pandas as pd
titanic = sns.load_dataset('titanic') #loading the dataset
print(titanic.head()) #print the output
```

```python
#Q11
titanic = sns.load_dataset('titanic') #loading the dataset from seaborn library
survivors_pclass3 = len(titanic[(titanic['pclass'] == 3) & (titanic['survived'] == 1)]) #total number of survivors in the third class
print(survivors_pclass3, "survivors") #display the output
```

```python
#Q12
avg_age_female_survived = titanic[(titanic['sex'] == 'female') & (titanic['survived'] == 1)]['age'].mean() #average age of female passengers who survived
print(f"Average age of female passengers who survived: {avg_age_female_survived:.2f}") #display the output
```

```python
#Q13
children = titanic[titanic['age'] <= 18] #children age less than and equal to 18
survival_rate_children = children['survived'].mean() * 100 #survival rate of passengers aged 0-18 years
print(f"Survival rate of children: {survival_rate_children:.2f}%") #display the output
```

```python
#Q14
male_alone = titanic[(titanic['sex'] == 'male') & (titanic['sibsp'] == 0) & (titanic['parch'] == 0)] #survival rate of male passengers traveling alone
survival_rate_male_alone = male_alone['survived'].mean() * 100 #percentage of survival rate of male alone
print(f"Survival rate of male passengers traveling alone: {survival_rate_male_alone:.2f}%") #display the output
```

```python
#Q15
sibsp_passengers = titanic[titanic['sibsp'] > 0] #sibsp passengers survived
survival_rate_sibsp = sibsp_passengers['survived'].sum() #survival rate of passengers with siblings/spouses aboard
total_with_sibsp = len(sibsp_passengers)
survival_rate_sibsp_percentage = survival_rate_sibsp / total_with_sibsp * 100 #Passengers percentage of survival rate
print(f"Survival rate of passengers with siblings/spouses: {survival_rate_sibsp_percentage:.2f}%") #display the output
```

```python
#Q16
#a
#Summary statistics and data distribution for numerical and categorical variables.
print("Summary Statistics for Numerical Columns:")
print(titanic.describe())

print ("\n")

print("Summary Statistics for Categorical Columns:")
print(titanic.describe(include=['object', 'category']))
```

```python
#b
#Analysis of missing values and their potential impact.
#Check missing values
print("\n Missing Values in Each Column:")
print(titanic.isnull().sum())

#Visualization of missing values
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
sns.heatmap(titanic.isnull(), cmap='viridis', cbar=False)
plt.title('Heatmap of Missing Values')
plt.show()
```

```python
#c
#Visualization of key relationships (e.g., survival rate by class, age, gender, etc.).
# Survival rate by class
sns.barplot(x='pclass', y='survived', data=titanic)
plt.title('Survival Rate by Passenger Class')
plt.show()

# Age distribution of Survivors and Non-survivors
sns.histplot(data=titanic, x='age', hue='survived', multiple ='stack', kde=True)
plt.title('Age Distribution of Survivors and Non-survivors')
plt.show()

# Survival rate by gender
sns.barplot(x='sex', y='survived', data=titanic)
plt.title('Survival Rate by Gender')
plt.show()
```

```python
#d
#Identification of potential outliers or anomalies.
sns.boxplot(x=titanic['age']) #Boxplot to identify outliers in 'age' column
plt.title("Boxplot for Age Outliers")
plt.show() #Show the output
```
